```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Netflix_Cleaned.csv')

df.head()
```

```python
# Step 1: Extract and clean country information

country_data = df[['show_id', 'title', 'country']].copy()

country_data['country'] = country_data['country'].str.strip()

country_data.head()
```

```python
# Step 2: Calculate content count by country

country_counts = country_data['country'].value_counts()

country_counts.head(10)
```

```python
# Step 3: Identify top content-producing countries

top_countries = country_counts[country_counts.index != 'Not Given'].head(10)

top_countries
```

```python
# Step 4: Create chart and ranking

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title("Top 10 Countries by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

## Findings

- The United States has the highest number of Netflix titles, with 3,240 titles.
- India ranks second with 1,057 titles.
- The United Kingdom ranks third with 638 titles.
- The United States has significantly more Netflix titles than any other country in the top 10.
- The dataset shows a strong concentration of Netflix content among the leading countries, particularly the United States.
