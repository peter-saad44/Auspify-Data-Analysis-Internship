```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Netflix_Cleaned.csv')

df.head()
```

```python
# Step 2: Calculate yearly content releases

yearly_releases = df['release_year'].value_counts().sort_index()

yearly_releases
```

```python
# Step 3: Identify growth and decline trends

yearly_change = yearly_releases.diff()

yearly_change
```

```python
# Step 4: Visualize yearly release trends

plt.figure(figsize=(12, 6))

sns.lineplot(
    x=yearly_releases.index,
    y=yearly_releases.values,
    marker='o'
)

plt.title("Netflix Content Releases by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Findings

- The number of Netflix titles increased significantly over the years.
- Content production grew rapidly after 2000.
- The highest number of titles was released in 2018, with 1,146 titles.
- The number of releases decreased after 2018, reaching 592 titles in 2021.
- The data shows a clear expansion of Netflix content production, especially during the 2010s.
